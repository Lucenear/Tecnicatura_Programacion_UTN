## 1. Descripcion del Caso y Problema
La organizacion hotelera busca comprender el fenomeno de las cancelaciones de reservas. El problema general que orienta este analisis es identificar patrones y factores que influyen en que un cliente cancele su reserva antes de la fecha de llegada.

Comprender esto es crucial porque las cancelaciones generan incertidumbre en la ocupacion, perdida de ingresos potenciales y dificultades en la planificacion de recursos (personal, alimentos, mantenimiento). Anticiparlas permite al hotel tomar decisiones proactivas.


## 2. Variable Objetivo: is_canceled
Dentro del dataset, la variable `is_canceled` es la columna fundamental para este analisis.
- Representa el estado final de la reserva.
- Valor 0: La reserva NO fue cancelada (el cliente se hospedo).
- Valor 1: La reserva SI fue cancelada (el cliente no llego).

Esta variable nos permite clasificar cada registro y es lo que intentaremos predecir o explicar en funcion de otras caracteristicas (como el tiempo de anticipacion, el tipo de habitacion, el pais de origen, etc.).

### Variables Categoricas
Observamos la frecuencia de las categorias principales para entender la composicion del negocio.

```python
# Frecuencia de cancelaciones (Variable Objetivo)
print("--- Distribucion de is_canceled ---")
print(df['is_canceled'].value_counts(normalize=True) * 100)

print("\n--- Distribucion por Hotel ---")
print(df['hotel'].value_counts())

print("\n--- Distribucion por Tipo de Deposito ---")
print(df['deposit_type'].value_counts())
```

**Observaciones de las categoricas:**
- **is_canceled:** Se observa que una proporcion muy significativa de las reservas en este dataset terminan siendo canceladas (el valor 1 tiene una alta frecuencia). Esto confirma que es un problema relevante para el hotel.
- **hotel:** Hay una distribucion entre City Hotel y Resort Hotel que permitira hacer comparativas interesantes mas adelante.
- **deposit_type:** La mayoria abrumadora de las reservas son "No Deposit". Habra que analizar si las pocas reservas "Non Refund" o "Refundable" se comportan distinto respecto a las cancelaciones.

## 7. Consideraciones eticas

El uso de este dataset implica ciertas responsabilidades, incluso en esta etapa temprana:

- **Privacidad:** Aunque el dataset parece anonimizado (no hay nombres reales de clientes, solo codigos de pais o agentes), representa comportamientos reales de personas.
- **Sesgos en los datos:** Los datos reflejan las politicas del hotel en un periodo especifico. Si el hotel tuvo una promocion agresiva de "No Deposit" en esos años, los datos estaran sesgados hacia ese comportamiento. Las conclusiones no necesariamente aplican a otros hoteles o contextos economicos.
- **Limites de la conclusion (Correlacion vs Causalidad):** En esta exploracion vemos que ciertas variables se mueven junto con las cancelaciones (por ejemplo, mayor lead_time, mayor cancelacion). Debemos recordar que esto es una correlacion; el tiempo de anticipacion no "causa" magicamente la cancelacion, sino que ambos estan ligados a la planificacion del cliente.
